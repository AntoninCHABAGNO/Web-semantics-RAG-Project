```python
from rdflib import Graph

tbox = Graph()
tbox.parse("data/ontologie_tbox.ttl")

expanded_kb = Graph()
expanded_kb.parse("data/expanded_kb.ttl")

abox = Graph()
abox.parse("data/queens_gambit_graph_abox.ttl")

g = tbox + abox + expanded_kb # si tu veux garder un graphe total aussi

print("TBox triples:", len(tbox))
print("ABox triples:", len(abox))
print("expanded_kb triples:", len(expanded_kb))
print("Total triples:", len(g))
subjects = set()
predicates = set()
objects = set()

for s, p, o in g:
    subjects.add(s)
    predicates.add(p)
    objects.add(o)

print("Total Subjects:", len(subjects))
print("Total Predicates:", len(predicates))
print("Total Objects:", len(objects))
print("Total Triples:", len(g)) 
```
