```python
from rdflib import Graph

tbox = Graph()
tbox.parse("data/ontologie_tbox.ttl")

abox = Graph()
abox.parse("data/queens_gambit_graph_abox.ttl")

g = tbox + abox  # si tu veux garder un graphe total aussi

print("TBox triples:", len(tbox))
print("ABox triples:", len(abox))
print("Total triples:", len(g))
```

```python
subjects = set()
predicates = set()
objects = set()

for s, p, o in abox:
    subjects.add(s)
    predicates.add(p)
    objects.add(o)

print("ABox Subjects:", len(subjects))
print("ABox Predicates:", len(predicates))
print("ABox Objects:", len(objects))
print("ABox Triples:", len(abox)) 
```

```python
from collections import Counter

pred_count = Counter()

for s, p, o in abox:
    pred_count[p] += 1

def short(uri):
    return str(uri).split("#")[-1]

for p, c in pred_count.most_common(25):
    print(short(p), c) 
```

```python
query = """
PREFIX rdf: <http://www.w3.org/1999/02/22-rdf-syntax-ns#>

SELECT ?type (COUNT(?s) as ?count)
WHERE {
    ?s rdf:type ?type .
}
GROUP BY ?type
ORDER BY DESC(?count)
"""

for row in abox.query(query):
    print(short(row[0]), int(row[1]))
```

```python
from rdflib.namespace import RDF, RDFS

for i, (s, p, o) in enumerate(abox):
    if p in (RDF.type, RDFS.label):
        continue
    print(short(s), short(p), short(o))
    if i > 30:
        break
```
